```python

import numpy as np
import matplotlib.pyplot as plt

#given values
days = np.array([90, 110, 150])
input = np.array([15, 35, 18]) 
k = 0.023 

#function for m gesamt
def m_ges(t):
    t = np.asarray(t, dtype=float)

    return sum(m * np.exp(-k * (t - tag)) * (t >= tag)for tag, m in zip(days, input))

# timeline
t = np.linspace(80, 220, 2000)

# plotting
plt.figure(figsize=(10, 5.5))

#plot gesamt
plt.plot( t, m_ges(t), linewidth=2.5, label="Gesamtmasse")

# mark with scatter where pesticide is put into see
for day, inp in zip(days, input):
    after = float(m_ges(day))
    before = after - inp

    plt.vlines(day, before, after, linestyles="--")

    plt.scatter(day, after)

    plt.annotate(f"{after:.2f} g", (day, after), xytext=(7, 9), textcoords="offset points")

# label, title etc
plt.xlabel("Zeit t [Tage]")
plt.ylabel("Gesamtmasse m [g]")
plt.title("Exponentieller Zerfall mit Massenzugaben")

plt.xlim(80, 220)
plt.ylim(0, 55)

plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Aufgabe a)

#generate x axis
x = np.linspace(-10, 10, 500)

#define the funtinos
f = -4 * x**2 + 20
g = x**2

plt.figure(figsize=(8, 5))
#plot f
plt.plot(x, f, color="blue", linestyle=":",label=r"$f(x) = -4x^2 + 20$")
#plot g
plt.plot(x, g, color="green", linestyle="-",label=r"$g(x) = x^2$")

#labels
plt.xlabel("x")
plt.ylabel("y: Value of function")
#title
plt.title("Quadratic function")
#legende
plt.legend()
#layout
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# b) Intersection berechnen
def intersection(m1, b1, m2, b2):

#check for special conditions
    if m1 == m2:
        if b1 == b2:
            raise ValueError("Same functions. No intersection")
        else:
            raise ValueError("DFunctions are parallel. No intersecton")
        return None

    #insert straight equations and solve for x
    xs = (b2 - b1) / (m1 - m2)
    ys = m1 * xs + b1

#print that and shorten the number
    print(f"Intersection: S({xs:.6f} | {ys:.6f})")

#x plotbereich erstellen
    x = np.linspace(xs - 5, xs + 5, 500)

    plt.figure(figsize=(8, 5))
    #plot funtction 1
    plt.plot(x, m1 * x + b1, color="blue",label=f"f(x) = {m1:g}x + ({b1:g})")
    #plot function 2
    plt.plot(x, m2 * x + b2, color="red",label=f"g(x) = {m2:g}x + ({b2:g})")
    #plot intersection
    plt.scatter(xs, ys, color="green", s=70, zorder=3,label=f"S({xs:.3f} | {ys:.3f})")

#label, title, legend etc
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title("Schnittpunkt zweier Geraden")
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    return xs, ys

#call intersectiomn
intersection(0.5, 2, -1, 0)
```
