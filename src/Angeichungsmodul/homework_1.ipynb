## Drei Systeme:
1) Wald:

Identität: O<sub>2</sub>Produktion

Struktur: Substrat -> Bäume -> Insekten -> Räuber -> Destruenten -> Substrat (Wirkungsbeziehungen von Spezies) 

Konsistenz: Wenn Bäume aus dem System herausgenommen werden, wird kein O^2 produziert

2) Computer:

Identität: Berechnungshilfen für Menschen

Struktur: Powerunit -> CPU -> RAM -> GPU -> Display (Wirkungsbeziehungen von PC-Parts)

Konsistenz: Ohne CPU können keine Berehcnung durchgeführt werden

3) Visuelles System im Gehirn

Identität: Sehen, Dinge erkennen und verorten.

Struktur: Retina -> LGN ->  ventral and dorsal Stream ( with interconnections) -> higher cortical layers -> connections back to lower areas.

Konsistenz: Remove one component -> no Vision

## Drei Nicht-Systeme
1) Stein:
Identität: Kein Zweck, noch Verhalten
Struktur: Solange nicht auf molekularer Ebene betrachtet, keine multiple Objekte und damit keine Struktur
Konsistenz: Ohne Struktur und Identität keine Struktur

2) Schwarzes Loch:
Identität: Informationslöschung
Struktur: Keine, infinitesimaler kleiner Punkt im Raum, daher keine Strukutr
Konsistenz: Da keine Struktur, kann nichts weg genommen werden. 

3) Sandhaufen:
Identität: Sandvorrat
Struktur: Anordnung von Sandkörnern
Konsistenz: Wenn man Sand entfernt ist es immer noch ein Sandvorrat. also keine Konsistenz.

```python
#c)
def find_int_div(x):#
    i=1
    int_divs=[]
    while i <=x and len(int_divs)<15:
        if x % i == 0:
            int_divs.append(i)
        i+=1
    return int_divs
print("Int Divs of 110:", find_int_div(110), "\nInt Divs of 120:", find_int_div(1200))
```
