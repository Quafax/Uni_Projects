```python
#imports
import numpy as np
import matplotlib.pyplot as plt

#euler approximation
def euler(t_end, y_0, f, h):
    """
    t_end is maximal steps, y_0 is anfangsbedingung, f is function, h is schrittweite
    """
    #array for t
    ts = np.arange(0, t_end + h/2, h)
    #array for ys
    ys = np.empty_like(ts, dtype=float)
    #fill in anfangsbedingung
    ys[0] = y_0

    #alle ts durchgehen und mit euler aproximation formel nächstes y ausrechnen
    for i in range(len(ts) - 1):
        ys[i + 1] = ys[i] + h * f(ys[i], ts[i])
    #returnen
    return ts, ys

#function definieren
def f(y, t):
    return -0.5 * y

#analytische lösung
def analytic_solution(t, C):
    return C * np.exp(-0.5 * t)

#nur ausführen wenn direkt ausgeführt
if __name__ == "__main__":
    #anfangsbedingung
    y0 = 10

    #euler für verschiedene hs
    ts1, y1 = euler(10, y0, f, 0.1)
    ts2, y2 = euler(10, y0, f, 0.5)
    ts3, y3 = euler(10, y0, f, 1)

    #beide diagramme als subplots anzeigen
    ts_exact = np.linspace(1.5, 2.5, 500)
    fig, (ax1, ax2) = plt.subplots(2, 1)

    ax1.plot(ts1, y1, label="h=0.1")
    ax1.plot(ts2, y2, label="h=0.5")
    ax1.plot(ts3, y3, label="h=1")
    ax1.set_xlabel("t")
    ax1.set_ylabel("y(t)")
    ax1.set_title("Euler-Verfahren")
    #comment taht for full functin
    ax1.set_xlim(1.5, 2.5)
    ax1.set_ylim(2.5, 5)
    ax1.grid(True)
    ax1.legend()

    ax2.plot(ts_exact, analytic_solution(ts_exact, y0), label="y = C * e^(-0.5t)")
    ax2.set_xlabel("t")
    ax2.set_ylabel("y(t)")
    ax2.set_title("Analytische Funktion")
    #comment that for full function
    ax2.set_xlim(1.5, 2.5)
    ax2.set_ylim(ax1.get_ylim())
    
    ax2.grid(True)
    ax2.legend()

    fig.tight_layout()
    plt.show()
```

 Es ist zu erkennen, dass je kleiner das h wird, desto genauer näher sich die approximation durch euler an die analytische Lösung an.
