```python
def calc_temp():
    try:
        eingabe = input("Temperatur: ")
        #komma und punkt wegen english/deutsch unterschied
        temp = float(eingabe.replace(",", "."))
    except ValueError:
        print("Bitte eine gültige Zahl eingeben.")
        return
    #try to get the temp as input. if not valid, return and raise error

    if not (float("-inf") < temp < float("inf")):
        print("Bitte eine endliche Zahl eingeben.")
        return
    #check if value is ok

    #get if Celius or kelvin, put alwys in upper Case for safety, could add "Celius" and "Kelvin"
    einheit = input("Einheit (C oder K): ").strip().upper()

    #a bit unnessecary to use match here, but its introduced in 3.10 and i wanted to use it :D
    match einheit:
        case "C":
            if temp < -273.15:
                print("Die Temperatur darf nicht unter -273,15 °C liegen.")
    #weil obvioulsy nicht unter absoluter nullpunkt
                return
#needed to use f strings for formating
            print(f"Ergebnis: {temp + 273.15:.2f} K")

        case "K":
            if temp < 0:
                print("Die Temperatur darf nicht unter 0 K liegen.")
    #weil obvioulsy nicht unter absoluter nullpunkt
                return

            print(f"Ergebnis: {temp - 273.15:.2f} °C")
        case _:
            print("Bitte nur C oder K eingeben.")


def faculty(n):
    #check correct number
    if type(n) is not int or n < 0:
        raise ValueError("Bitte eine nichtnegative ganze Zahl übergeben.")
    #check trivial cases
    elif n == 0 or n == 1:
        return 1
    #calc non trivial cases
    else:
        result = 1
        for i in range(2, n + 1):
            result *= i

    #give result
        return result
    
if __name__ == "__main__":
    #activate this bottom line to use the temperature calculator
    #calc_temp()
    print("12! =", faculty(12))
```
