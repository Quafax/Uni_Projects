```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Western_rock_lobster.csv")
E = data["f"]
Y = data["Y"]
plt.scatter(E, Y)
plt.xlabel("Effort (million pot lifts)")
plt.ylabel("Yield (kt)")
plt.show()
```

```python

from scipy.optimize import curve_fit

def yield_model(E, a, b):
    return a * E - b * E**2

parameters, covariance = curve_fit(yield_model, E, Y)
a, b = parameters

print("a:", a)
print("b:", b)
```

```python
import numpy as np

E_curve = np.linspace(0, E.max(), 300)

plt.figure(figsize=(8, 5))
plt.scatter(E, Y, label="Observed data", alpha=0.7)
plt.plot(
    E_curve,
    yield_model(E_curve, a, b),
    color="red",
    label="Equilibrium model fit"
)

plt.xlabel("Effort (million pot lifts)")
plt.ylabel("Yield (kt)")
plt.legend()
plt.tight_layout()
plt.show()
```
